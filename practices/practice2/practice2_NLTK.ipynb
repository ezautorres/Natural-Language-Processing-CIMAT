### **Concordancia**

Podemos imprimir la concordancia de los términos:

```python
text1.concordance("monstrous") # search for a word in the text and show the context.
```

Y esto despliega el contexto de esta palabra en todo el documento de texto (Moby Dick) que fue encontrada 11 veces. Esto se llama **hipótesis distribucional:** Podemos modelar la semántica y decir que una palabra es similar a otra si tiene como contexto a palabras similares, i.e., interpretar las palabras por los tokens vecinos que tiene.

El texto 5 es gracioso ya que son textos de chats:

```python
text5.concordance("fuck") # search for a word in the text and show the context.
```

## **Distribución de Frecuencias**

```python
from nltk import FreqDist
fdist1 = FreqDist(text1)
vocabulary = fdist1.keys()
print(vocabulary)
print("whale aparece:", fdist1['whale'], "veces")
```

```python
fdist1.plot(50, cumulative = True)
```

## **Colocaciones:**

Una colocación es una expresión consistiendo de una o más palabras que corresponden a alguna forma convencional de decir las cosas

```python
text4.collocations()
```

¿Cómo saber cuál es la distribución de la longitud de las palabras dentro de un texto?

```python
setOFLenghts = [len(w) for w in text1]
fdist = FreqDist(setOFLenghts)
print(fdist)
print(fdist.keys())
fdist.plot(19, cumulative = False)
```
